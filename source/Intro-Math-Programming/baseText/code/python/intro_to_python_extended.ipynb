## 🐍 Dasar-Dasar Python

Python adalah bahasa pemrograman tingkat tinggi yang diinterpretasikan dan dikenal karena keterbacaan serta keserbagunaannya.

### Variabel dan Tipe Data
Python bertipe dinamis, sehingga Anda tidak perlu mendeklarasikan tipe variabel secara eksplisit.

- **Bilangan bulat**: Bilangan tanpa bagian pecahan (misalnya 10, -5)
- **Bilangan titik mengambang**: Bilangan desimal (misalnya 3.14, -2.7)
- **String**: Urutan karakter (misalnya "Hello", "Alice")
- **Boolean**: True atau False


```python
# Basic variable assignment
x = 10           # Integer
y = 3.14         # Float
name = "Alice"   # String
is_happy = True  # Boolean

print(x, y, name, is_happy)
```

### Operasi Dasar

Python mendukung operasi aritmetika umum:
- `+` (penjumlahan)
- `-` (pengurangan)
- `*` (perkalian)
- `/` (pembagian)
- `**` (perpangkatan)
- `//` (pembagian bulat ke bawah)
- `%` (modulus)



```python
# Demonstration of basic arithmetic
a = 15
b = 4

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
```

### Koleksi: List, Tuple, Dictionary, dan Set

Berikut adalah struktur data mendasar dalam Python:

- **List** adalah urutan yang dapat diubah (misalnya `[1, 2, 3]`).
- **Tuple** adalah urutan yang tidak dapat diubah (misalnya `(10, 20, 30)`).
- **Dictionary** menyimpan pasangan kunci dan nilai (misalnya `{"apple": 2, "banana": 3}`).
- **Set** menyimpan elemen unik (misalnya `{1, 2, 3}`).


```python
my_list = [1, 2, 3, 4]              # List
my_tuple = (10, 20, 30)              # Tuple
my_dict = {"apple": 2, "banana": 3}  # Dictionary
my_set = {1, 2, 2, 3, 4}             # Set (duplicates are removed)

print("List:", my_list)
print("Tuple:", my_tuple)
print("Dictionary:", my_dict)
print("Set:", my_set)

# List operations
my_list.append(5)
print("\nAfter appending 5 to my_list:", my_list)

# Dictionary operations
my_dict["cherry"] = 5
print("\nAfter adding cherry to my_dict:", my_dict)
```

### Pengindeksan dan Pengirisan

List, tuple, dan string mendukung pengindeksan (untuk mengakses satu elemen) dan pengirisan (untuk mengakses subrentang elemen).

list_or_string[start:end:step]
- `start`: Indeks awal (nilai bawaan 0)
- `end`: Indeks akhir (tidak disertakan; nilai bawaan adalah panjang urutan)
- `step`: Ukuran langkah (nilai bawaan 1)


```python
# Indexing and slicing
colors = ["red", "green", "blue", "yellow", "purple"]
print("colors[0]:", colors[0])
print("colors[1:3]:", colors[1:3])
print("colors[:3]:", colors[:3])
print("colors[::2]:", colors[::2])
```

### Alur Kendali

Python menggunakan indentasi untuk menetapkan blok. Pernyataan utamanya meliputi:
- **if**, **elif**, dan **else** untuk percabangan bersyarat
- Perulangan **for** untuk mengiterasi urutan
- Perulangan **while** untuk eksekusi berulang selama suatu kondisi terpenuhi


```python
# Using a for loop to iterate over a dictionary
for fruit in my_dict:
    print(f"{fruit} has {my_dict[fruit]} items.")

# Example of if-else
num = 7
if num % 2 == 0:
    print(f"{num} is even.")
else:
    print(f"{num} is odd.")
```

### Fungsi

Gunakan `def` untuk menetapkan blok kode yang dapat digunakan kembali. Fungsi dapat menerima parameter dan mengembalikan nilai.


```python
def square(x):
    return x * x

val = 5
print(f"The square of {val} is {square(val)}.")
```

## 🔢 NumPy: Komputasi Numerik dalam Python

NumPy adalah paket dasar untuk komputasi numerik dalam Python. Paket ini menyediakan:
- Objek larik berdimensi N yang andal
- Fungsi untuk melakukan operasi tervektorisasi
- Alat untuk aljabar linear, pembangkitan bilangan acak, dan banyak lagi


```python
import numpy as np

# Create a 1D array
a = np.array([1, 2, 3, 4, 5])
print("Array a:", a)

print("Mean:", np.mean(a))
print("Standard Deviation:", np.std(a))
print("Element-wise multiplication:", a * 2)
```

### Larik Multidimensi
NumPy mendukung larik multidimensi (matriks, tensor). Mari kita buat larik 2D (matriks) dan jelajahi beberapa operasinya.


```python
# 2D array (matrix)
b = np.array([[1, 2],
              [3, 4]])
print("Matrix b:\n", b)

# Transpose
print("\nTranspose of b:\n", b.T)

# Shape
print("\nShape of b:", b.shape)

# Reshaping
c = np.array([1,2,3,4,5,6])
c_reshaped = c.reshape((2, 3))
print("\nOriginal c:", c)
print("Reshaped c to (2,3):\n", c_reshaped)
```

### Pembangkitan Data Acak

NumPy menyediakan cara praktis untuk membangkitkan data acak, yang sering berguna bagi simulasi atau pengujian.


```python
# Generating random data
np.random.seed(42)  # for reproducibility

rand_array = np.random.rand(3, 3)  # 3x3 array of random numbers in [0,1)
print("Random 3x3 array:\n", rand_array)

rand_ints = np.random.randint(0, 10, size=(2, 5))  # 2x5 array of random integers in [0,10)
print("\nRandom 2x5 array of integers:\n", rand_ints)
```

## 🐼 Pandas: Manipulasi Data yang Mudah

Pandas adalah pustaka yang andal untuk analisis dan manipulasi data. Pustaka ini memperkenalkan dua struktur data utama:

- **Series**: Larik 1D berlabel  
- **DataFrame**: Tabel 2D berlabel (seperti lembar sebar)

Pandas memudahkan pemuatan, pembersihan, dan analisis data dari berbagai sumber, termasuk berkas CSV, berkas Excel, basis data SQL, dan banyak lagi.


```python
import pandas as pd

# Creating a simple DataFrame from a dictionary
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "Score": [85.0, 92.5, 88.0]
}
df = pd.DataFrame(data)
df
```

### Operasi Dasar DataFrame

DataFrame Pandas memiliki banyak metode dan properti bawaan yang memungkinkan manipulasi data secara cepat.


```python
# Check the average score
print("Average Score:", df["Score"].mean())

# Sort by age
print("\nSorted by Age:")
print(df.sort_values("Age"))

# Accessing columns
print("\nAccessing the 'Name' column:")
print(df["Name"])

# Filtering rows
print("\nFiltering rows where Age > 25:")
print(df[df["Age"] > 25])
```

### Pengelompokan dan Agregasi

Pengelompokan data memungkinkan Anda membagi data menjadi beberapa kelompok berdasarkan kriteria tertentu, menerapkan fungsi pada setiap kelompok, lalu menggabungkan hasilnya.


```python
# Create a slightly bigger DataFrame for demonstration
data_extended = {
    "Name": ["Alice", "Bob", "Charlie", "Alice", "Bob", "Charlie"],
    "Category": ["A", "A", "B", "B", "A", "B"],
    "Value": [10, 20, 5, 15, 30, 25]
}
df_ext = pd.DataFrame(data_extended)
print(df_ext)

# Group by Category and sum the Value
grouped = df_ext.groupby("Category")["Value"].sum()
print("\nSum of Value by Category:")
print(grouped)
```

## 📁 Mengimpor Data Nyata

Mari kita impor himpunan data dunia nyata dari berkas CSV yang tersedia daring. Kita akan menggunakan data perjalanan udara.

> **Catatan**: Jika Anda mengalami masalah ketika memuat data dari URL (misalnya karena pembatasan jaringan), Anda dapat mengunduh berkas tersebut dan memuatnya secara lokal.


```python
# Load dataset from URL
url = "https://people.sc.fsu.edu/~jburkardt/data/csv/airtravel.csv"
df_air = pd.read_csv(url)
df_air.head()
```

```python
# Checking structure
df_air.info()
```

### Mengiris dan Memfilter DataFrame

Dalam Pandas, Anda dapat memfilter DataFrame dengan menerapkan kondisi Boolean pada salah satu kolomnya.
Misalnya, jika Anda hanya menginginkan baris dengan nilai pada kolom numerik tertentu lebih besar daripada ambang, gunakan:

`df[df["column_name"] > some_value]`

Anda dapat menggabungkan beberapa kondisi menggunakan operator bit-per-bit `&` (DAN) dan `|` (ATAU) beserta tanda kurung:

`df[(df["colA"] > val1) & (df["colB"] == val2)]`

Berikut kami tampilkan contoh praktis.


```python
# Example DataFrame
df_filter_example = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "Age": [25, 19, 34, 42, 17],
    "Score": [85.0, 92.5, 88.0, 75.0, 95.0]
})

print("Original DataFrame:")
display(df_filter_example)

# Filter rows where Age is greater than 20
filtered_df = df_filter_example[df_filter_example["Age"] > 20]
print("\nFiltered DataFrame (Age > 20):")
display(filtered_df)
```

## 📊 Matplotlib: Visualisasi Dasar

Matplotlib adalah pustaka untuk membuat visualisasi dalam Python. Pustaka ini menawarkan beragam plot dan pilihan penyesuaian.

### Plot Garis Dasar

Kita akan menggunakan data perjalanan udara untuk membuat plot garis.


```python
import matplotlib.pyplot as plt

months = df_air["Month"]
totals = df_air.iloc[:, 1:].sum(axis=1)

plt.figure(figsize=(8, 4))
plt.plot(months, totals, marker='o', linestyle='-')
plt.title("Total Air Travel Per Month")
plt.xlabel("Month")
plt.ylabel("Total Passengers")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Diagram Batang
Diagram batang sangat cocok untuk data kategoris atau untuk membandingkan beberapa kategori.


```python
plt.figure(figsize=(8, 4))
plt.bar(months, totals)
plt.title("Total Air Travel Per Month (Bar Chart)")
plt.xlabel("Month")
plt.ylabel("Total Passengers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Plot Sebar
Jika Anda memiliki dua variabel kontinu, plot sebar membantu memvisualisasikan hubungannya. Berikut contoh singkat dengan data acak.


```python
x_rand = np.random.rand(50)
y_rand = np.random.rand(50)

plt.figure(figsize=(8, 4))
plt.scatter(x_rand, y_rand)
plt.title("Random Scatter Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.tight_layout()
plt.show()
```

### Histogram
Histogram membantu memvisualisasikan distribusi data numerik.


```python
data_for_hist = np.random.randn(1000)  # 1000 normally distributed points

plt.figure(figsize=(8, 4))
plt.hist(data_for_hist, bins=20)  # 20 bins
plt.title("Histogram of 1000 Random Values")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
